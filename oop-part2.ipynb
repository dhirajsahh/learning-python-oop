Write OOP classes to handle the following scenarios:
A user can create and view 2D coordinates
A user can find out the distance between 2 coordinates
A user can find find the distance of a coordinate from origin
A user can check if a point lies on a given line
A user can find the distance between a given 2D point and a given line

```python
class Point:

    def __init__(self,x,y):
        self.x_cord=x
        self.y_cord=y
    def __str__(self):
        return "<{},{}>".format(self.x_cord,self.y_cord)

    def distance(self,other):
        distance=(self.x_cord-other.x_cord)**2+(self.y_cord-other.y_cord)**2
        return distance**0.5
    def distance_from_origin(self):
        return self.distance(Point(0,0))

class Line:
    def __init__(self,A,B,C):
        self.A=A
        self.B=B
        self.C=C
    def __str__(self):
        return "{}x+{}y+{}=0".format(self.A,self.B,self.C)
    def check_point(self,point):
        return self.A*point.x_cord+self.B*point.y_cord+self.C==0

    def shortest_distance(Line,Point):
        deno=(Line.A**2+Line.B**2)**0.5
        return round(abs(Line.A*Point.x_cord+Line.B*Point.y_cord+Line.C)/deno,2)
    def check_line_intersect(self,other):
        if self.A/other.A!=self.B/other.B:
            print("The line intersect at one point")
            x=(self.B*other.C-other.B*self.C)/(self.A*other.B-other.A*self.B)
            y=(self.C*other.A-other.C*self.A)/(self.A*other.B-other.A*self.B)
            return Point(x,y)

            


p1=Point(3,4)
p2=Point(1,12)
p1.distance(p2)
p2.distance_from_origin()
l1=Line(1,1,-2)
l1.check_point(p2)
l1.shortest_distance(p2)
l2=Line(2,3,4)
print(l2.check_line_intersect(l1))


        
```

## Mutablity of object


```python
class Person:
    def __init__(self):
        self.name="Dhiraj"
        self.gender="male"
def greet(person):
    person.name='rahul'
    return person

p=Person()
print(id(p))  #125737934307824
p1=greet(p)
print(p1.name)
print(id(p1))  #125737934307824

# here we can see both are pointing to same location after modification they are 
# pointing to same memory location and hence we can say they are immutable 
# if they were immutable it points to another memory locations 
```

```python
def increment(a):
    a+= 1  # Creates a new integer object; does not mutate the original
    return a

a = 10
print(id(a))  #125738290709008
x=increment(a)
print(id(x))  #125738290709040 
# here both are different because the int variable is immutable 
```

1. Mutable Objects (Lists, Dictionaries, Sets, Custom Objects)
If you modify the contents of a mutable object in-place, the change reflects outside the function because both variables point to the same object.

2. Immutable objects cannot be modified in-place. Any operation that appears to modify them actually creates a new object in memory, leaving the caller's variable untouched

## Encapsulation


```python
class Test:

    def __init__(self):
        self.name="Dhiraj"
        self.totalmarks=650

    def take_average(self):
        return self.totalmarks/8

t1=Test()
t1.totalmarks="abc"
t1.take_average()
#This is gives error
# here we are changing the attribute of the class outside the class
#in order to make private we use __in front of atrribute or method eg.__totalmarks syntax __variablename
```

```python
class Test:

    def __init__(self):
        self.__name="Dhiraj"
        self.__totalmarks=650

    def take_average(self):
        return self.__totalmarks/8

t1=Test()
t1.totalmarks="abc"  ## This is like creating a new attribute to a object outside the class 
                      #it is not modifying my private variable
print(t1.totalmarks)
t1.take_average() #works and doesnot give error 

#now another thing 
t1.__totalmarks=100
print(t1.__totalmarks)  #100
print(t1.take_average()) #81.25 -->the actaull one 650/8abc

# hence we can conclude that it is same like as creating the newattribute to a object outside the class
# with name __totalmarks . this is not modifying the the private variable named __total marks 
# and now we understand why it is so.
```
