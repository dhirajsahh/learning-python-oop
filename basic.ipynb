```python
class ATM:

    #constructor -->special function -->that get automatically called when obj is created

    def __init__(self):
        self.pin=0
        self.balance=0
        self.menu() #doing this automaticaylly call the menu when ever obj is created
    def menu(self):
        user_input=input("""
          Hi how can i help you?
          1.Press 1 to create pin
          2.Press 2 to change pin
          3.Press 3 to check balance
          4.Press 4 to withdraw
          5.Anything else to exit
        """)
        if user_input=='1':
            self.create_pin()
            self.menu()
        elif user_input=='2':
            self.change_pin()
            self.menu()
        elif user_input=='3':
            self.check_balance()
            self.menu()
        elif user_input=='4':
            self.withdraw()
            self.menu()
        else:
            exit
        
    def create_pin(self):
        userpin=input("enter your pin")
        self.pin=userpin
        userbalance=int(input("enter balance"))
        self.balance=userbalance
        print("pin created successfully")
    def change_pin(self):
        currentpin=input("Enter the current pin")
        if currentpin==self.pin:
            new_pin=input("Enter the new Pin")
            self.pin=new_pin
            print("your pin is successfully changed")
        else:
            print("Your pin is incorrect")
    def check_balance(self):
        userpin=input("Enter your pin")
        if userpin==self.pin:
            print("your balance is ",self.balance)
        else:
            print("Your pin is incorrect")
    def withdraw(self):
        userpin=input("Enter your pin")
        if userpin==self.pin:
            currentwithDraw=int(input("Enter the amount to draw to withdraw"))
            if currentwithDraw<=self.balance:
                self.balance=self.balance-currentwithDraw
                print("your balance is successfully withdrawn")
            else:
                print("your balance is lesser")
        else:
            print("Your pin is incorrect")
            
```

```python
# fraction class
class Fraction:
    def __init__(self,x,y):  #parameterized constructor it is one which take input
        self.num=x
        self.deno=y
    
    def __str__(self): # the second magic method 
        return "{}/{}".format(self.num,self.deno)

    def __add__(self, other):
        new_num=self.num*other.deno+self.deno*other.num
        new_deno=self.deno*other.deno
        return "{}/{}".format(new_num,new_deno)
    def __sub__(self,other):
        new_num=self.num*other.deno-self.deno*other.num
        new_deno=self.deno*other.deno
        return "{}/{}".format(new_num,new_deno)
    def __mul__(self,other):
        new_num=self.num*other.num
        new_deno=self.deno*other.deno
        return "{}/{}".format(new_num,new_deno)
    def __truediv__(self,other):
        new_num=self.num*other.deno
        new_deno=self.deno*other.num
        return "{}/{}".format(new_num,new_deno)

    #non magic method
    def convert_to_decimal(self):
        return self.num/self.deno
```

# second magic method : __str__(self) when we call print and pass our object it automatically call this __str__(self)

```python
fr1=Fraction(3,4)
fr2=Fraction(1,2)

```
