```python
#Aggregation Has a relationship -->means one class owns the other
#e.g customer has address. Restuarant has a menu
class Customer:
    def __init__(self,name,gender,address):
        self.name=name
        self.gender=gender
        self.address=address
    def print_address(self): # this class cannot access the private member of the other class although we are passing as refrecne
        print(self.address.get_city(),self.address.pin,self.address.state) 
       # print(self.address._Address__city) this is still valid but we not do this . we use getter and setter
class Address:
    def __init__(self,city,pin,state):  
        self.__city=city
        self.pin=pin
        self.state=state
    def get_city(self):
        return self.__city
add1=Address("Brahampuri","1023","madhesh")


cust=Customer('Dhiraj','Male',add1)

cust.print_address()
        
```
